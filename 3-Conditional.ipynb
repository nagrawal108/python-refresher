```python
#Create a function
def categorize_age(age):
    if age < 13:
        return "Child"
    elif 13 <= age < 18:
        return "Teen"
    else:
        return "Adult"

# Example test
age_category = categorize_age(14)
print(age_category)
```

```python
def positive_negative(num):
    if num >= 0:
        if num==0:
            return "Zero"
        else:
            return "Positive number"
    else:
        return "Negative number"
        
#example test of calling the function
print("15 =>", positive_negative(15))
print("0 =>", positive_negative(0))
print("-4 =>", positive_negative(-4))
```

### Nested loops

```python
for i in range (2,11):
    print("Table of", i)
    for j in range(1, 11):
        print(i, "x", j, "=", i*j)
```

```python
li = [1, 2, 3]
li2 = [4,5,6]
li3 = []

#Indexing
for i in range(len(li)):
    print(li[i]+li2[i])
    li3.append(li[i]+li2[i])
    
print (li3)

#zip combines lists, tuples, dictionary but not sets because it is unordered
#lenght needs to be the same, otherwise zip cuts to the minimum lenght.
for i, j in zip(li, li2):
    print(i+j)
    
for i in zip(li, li2,li, (11, 22, 33, 44)):
    print(i)
    
print(zip(li, li2))
```

#### Break and Continue in loops

```python
#Break

for i in range(1, 11):
    print(i)
    if i == 8:
        break
```

```python
# Use of break statement inside the loop
for i in "Hello string":
    if i == "l":
        break
    print(i)
print("End of Loop")
```

```python
# Use of continue statement inside the loop
for i in "Hello string":
    if i == "l":
        continue
    print(i)
print("End of Loop")
```

```python
li = ["Nikhil", "Robert", "Shawn", "Shrirang"]
for i in li:
    if i == "Shawn":
        break
    print(i)
print("End of Loop")
```

```python
for i in range (2,11):
    print("Table of", i)
    for j in range(1, 11):
        if j == 4:
            break
        print(i, "x", j, "=", i*j)
```

```python
# Use of continue statement to only print l 
for i in "Hello string":
    if i == "l":
        print (i)
    else:
        continue
print("End of Loop")
```

```python
li = ["Nikhil", "Robert", "Shawn", "Shrirang"]
for i in li:
    if i.upper() == "SHAWN":
        break
    print(i)
else:    
    print("Else End of Loop")
print("End of loop")
```

## List Comprehension

```python
#This is a list comprehension
squares = [x**2 for x in range(1, 6)]
print(squares)
# Output: [1, 4, 9, 16, 25]
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
evens = [x for x in numbers if x % 2 == 0]
print(evens)
# Output: [2, 4, 6, 8]
```

```python
li = []
for i in range(6):
    li.append(i**2)
print(li)
```

```python
li = [x**2 for x in range(6) if x%2 !=0] #only odd numbers 
print(li)
```

```python
li = ["Odd" if x%2 !=0 else "Even" for x in range(6) ] #only odd numbers 
print(li)
print(len(li))
```

```python
import time
start = time.time()
li =[]
for i in range(1,6):
    li.append(i**2)
end = time.time()
py_time = end-start
print(li, py_time)

start = time.time()
li = [x**2 for x in range(1,6)]
end = time.time()
py_time2 = end-start
print(li, py_time2)
print(py_time-py_time2)
```

```python
###Dictionary Comprehension
#Creating a dictionary with numbers and their squares
squares_dict = {x: x**2 for x in range(1, 6)}
print(squares_dict)
# Output: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
#Filtering a dictionary to keep only even values
num_dict = {x: x**2 for x in range(10) if x % 2 == 0}
print(num_dict)
# Output: {0: 0, 2: 4, 4: 16, 6: 36, 8: 64}
```

```python
#Creating a set of squares
squares_set = {x**2 for x in range(1, 6)}
print(squares_set)
# Output: {1, 4, 9, 16, 25}
#Removing duplicates from a list using set comprehension
numbers = [1, 2, 2, 3, 4, 4, 5, 5, 6]
unique_numbers = {x for x in numbers}
print(unique_numbers)
# Output: {1, 2, 3, 4, 5, 6}
```
