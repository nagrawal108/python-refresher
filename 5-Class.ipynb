```python
class Counter:
   def __init__(self):
       self.count = 0
   def increment(self):
       self.count += 1
   def get_count(self):
       return self.count
c = Counter()
c.increment()
print(c.get_count()) # 1
c.increment()
print(c.get_count())
```

```python
#Class
class Dog:
    legs = 4 #Class attribute - default characteristics
    def __init__(self, breed, age, color):
        self.breed = breed #data attribute
        self.age = age #data attribute
        self.color = color
    
    def bark(self): #instance method as it has self word
        print(f"{self.breed} is barking")
        return True
    
    def sleep(self, hours): #instance method as it has self word with an argument
        print(f" {self.breed} is sleeping for {hours} hours") #formatted string
        return True

Max = Dog("Labrador", 3, "Brown") #create an object
print(Max.breed, Max.age, Max.color, Dog.legs)

Rock = Dog("Bull Dog", 4, "White")
print (Rock.breed, Rock.age, Rock.color, Rock.legs)

print(Rock.bark())
print(Rock.sleep(20))
Dog.legs = 3
print(Rock.legs)
```

## Encapsulation
#### Encapsulation binds data members and functions into a single unit
#### It hides the state of an object based on the type - public, _protected and __private

```python
class Person:
    def __init__(self, name, nationality,salary):
        self.name = name #public
        self._nationality = nationality #protected
        self.__salary = salary #private
    
    def showSalary(self):
        print(f"The salary of {self.name} is {self.__salary}")
        
Krishna = Person("Krishna", "Goloka", 900000000000000000000000000000000000)
print(Krishna.name, Krishna._nationality, Krishna._Person__salary) #Private attributes not recommended to access outside the class, using a technique called Name Mangling
#We can also access private attributes through getter and setter methods
print(Krishna.showSalary())
```

## Inheritance
### The process of forming a new class from am existing class or base class

```python
# Single Level Inheritance

class Parent_class:
    def parent(self):
        print("Hey I'm the parent class")
        
class Child_class(Parent_class):
    def child(self):
        print("Hey I'm the child class derived from the parent class")
        
obj = Child_class()

obj.parent()
obj.child()
```
