## File Handling in Python
### Create, read, write and close files
#### Used for data manipulation, storage & transfer


```python
#Create and open a file

# file = open('example.txt', 'w') #if file doesn't exist it will create it, if opened in write or append(a) mode
# file.close()
file = open ("example.txt", "r")
content = file.read() # read entire content of file
print(content)


#read one line at a time
line = file.readline()
print(line)

#reading all lines into a list
lines = file.readlines()
print(lines)

file.close()

file = open("example.txt", "a")

#write line or lines to the file
file.write("\n Hello, World! \n")
lines = ["First line\n", "Second line\n"]
file.writelines(lines)
file.close()

file = open ("example.txt", "r")
content = file.read()
print(content)

file.close() #close the file
```
