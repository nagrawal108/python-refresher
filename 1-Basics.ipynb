#### Sets, Lists, Tuples, Dictionary  => Mutable and Immutable

```python
set = {1,2,2,3, 3,4, "a", "b", "c"} #partially mutable - because we have certain limitations to mutable
print (set,type(set)) #sets are unordered and removes duplicates
```

```python
list = [1,2,2,3, 3,4, "a", "b", "c"] #mutable
print (list, type(list)) #ordered and shows everything in the list
```

```python
list[0]= "AI"
print (list, type(list)) #hence list is mutable as we were able to make changes to the list
```

```python
tuple = (1,2,2,3, 3,4, "a", "b", "c") #immutable as we can't change the tuple
print (tuple, type(tuple)) #can't replace elements in a tuple
```

```python
tuple[0] ="AI" #Can't do this and will throw an error
print(tuple)
```

```python
# Dictionary
dictionary = {"Key" : "Value", "Name" : "Nikhil", "Age": 44}
print(dictionary, type(dictionary))
```

```python
# Dictionary is mutable
dictionary ["Name"] = "Vivaan & Paarth"
dictionary ["Age"] = "14 & 12"
print (dictionary, type(dictionary))
```

```python
dictionary ["Education"] = "Masters"
print(dictionary)
```

```python
removed_value = dictionary.pop("Education")
print(dictionary)
print(removed_value)
```
