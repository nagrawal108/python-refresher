#### List - append, pop, extend, insert

```python
# append and pop and extend
fruits = ["apple", "banana", "cherry", "mango"]
print ("Before:", fruits) 
fruits.append("kiwi")
print("After:", fruits)
fruits.pop()
print("After pop:", fruits) #removes the last element
fruits.pop() 
print(fruits)
fruits.pop(1) #removes the element at the specified index
print(fruits)
fruits.extend(["pineapple", "watermelon"])
print(fruits)
fruits.insert(1, ["grape","blueberry"])
print(fruits)

```

### Sets 

```python
set1={1, 2, 3, 4, 5, 5 }
print(set1)
empty_set = set()

print(type(empty_set))
```

```python
#union of sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}
union_set = set1.union(set2)
print(union_set) #whatever is unique 
print(set1 | set2)
common_set = set1.intersection(set2)
print(common_set) #whatever is common between two sets
print(set1 & set2)
difference_set = set1.difference(set2)
print(difference_set)
print(set1-set2)
diff_set = set2.difference(set1)
print(diff_set, type(diff_set))
set3 = {5, (6,7,8), 6, 7} #tuples allowed within a list as tuples are immutable
diff_set1 = set3.difference(set2)
print(set3)
```

#### NumPy

```python
import numpy as np

# 1. Start with standard Python lists
house_sizes = [1500, 2000, 2500, 3000]  # Square feet
prices = [300000, 400000, 500000, 600000] # Dollars

# 2. Convert to NumPy Arrays (The AI-friendly format)
X = np.array(house_sizes)
y = np.array(prices)
print(X)
print(y)

# 3. Perform a vectorized operation (No "for" loops needed!)
# Let's say we want to normalize the data by dividing by 1000
X_normalized = X / 1000
print(f"Normalized Sizes: {X_normalized}")

# 4. Simple Linear Prediction: Price = Size * 200
prediction = X * 200
print(f"Predictions: {prediction}")
```
